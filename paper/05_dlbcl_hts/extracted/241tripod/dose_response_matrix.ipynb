```python
import pandas as pd
import numpy as np
```

```python
#create table like synergyfinder input
ncatsbtk = pd.read_csv("241/NCATSTable_btk.csv")
metadata = pd.read_csv("241/metadata.csv")
response = pd.read_csv("241/responses.csv")
```

```python
df_list = list()
for i in np.unique(metadata.BlockId):
    response1 = ncatsbtk.copy()
    meta = metadata[metadata.BlockId == i]
    rp = response[response.BlockId == i]
    response1.drug1 = response1.shape[0]*[meta.RowName.values[0]]
    response1.response = list(rp.Value)
    response1.block_id = response1.shape[0]*[i]
    df_list.append(response1)
```

```python
df = pd.concat(df_list)
```
